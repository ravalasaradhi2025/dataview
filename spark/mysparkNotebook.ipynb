```python
%sql
use catalog datamaster;
use schema bronze;


 
```

```python
%sql
create or replace table bronze.customer_reviews ( customer_id int, first_name string, last_name string, review string);

```

```python
%sql
insert into bronze.customer_reviews values (1, 'John', 'Smith', 'Great product!');
insert into bronze.customer_reviews values (2, 'Jane', 'Doe', 'Not so great product!');
insert into bronze.customer_reviews values (3, 'Mary', 'Jane', 'Great product!');
insert into bronze.customer_reviews values (4, 'Bob', 'Smith', 'Great product!');
insert into bronze.customer_reviews values (5, 'Alice', 'Doe', 'Not so great product!')
```

```python
%sql

select * from datamaster.bronze.customer_reviews;
```

```python
%sql
create or replace function datamaster.bronze.fullname_udf(first_name string, last_name string) 
returns string
return first_name || ' ' || last_name;
```

```python
%sql
select customer_id, datamaster.bronze.fullname_udf(first_name, last_name) as full_name, review from datamaster.bronze.customer_reviews;
    
```

```python
%sql
create or replace function datamaster.bronze.customer_analysis(review string)
returns string
return 
  case 
    when review like '%great%' then 'Positive'
    when review like '%not%' then 'Negative'
    else 'Neutral'
  end;
```

```python
%sql
select customer_id, review, datamaster.bronze.customer_analysis(review) as sentiment from datamaster.bronze.customer_reviews;
    
```
